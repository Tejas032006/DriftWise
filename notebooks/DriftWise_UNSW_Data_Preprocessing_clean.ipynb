# DriftWise: UNSW-NB15 data preprocessing
Run top to bottom. Result: two cleaned, chronologically sorted files saved in `MyDrive/DriftWise/data/processed/`:
- `UNSW_NB15_DriftWise_clean_chronological.csv.gz` (deduplicated, 640,788 rows): **the main real dataset**
- `UNSW_NB15_DriftWise_chronological_all_records.csv.gz` (700,001 rows): kept for a sensitivity check

Cell 2 asks you to upload the chronological raw file from your computer.

```python
# CELL 1: IMPORT LIBRARIES
import pandas as pd
import numpy as np
from google.colab import files
from IPython.display import display

pd.set_option("display.max_columns", 60)
```

```python
# CELL 2: UPLOAD RAW DATASET
uploaded = files.upload()
raw_filename = next(iter(uploaded))
print("Uploaded:", raw_filename)
```

```python
# CELL 3: LOAD DATA WITH AUTOMATIC HEADERS

# Read the dataset letting pandas automatically parse the header row
df = pd.read_csv(
    raw_filename,
    header=0,
    low_memory=False,
    encoding="latin1",
    compression="gzip"
)

print("Dataset shape:", df.shape)
print("Columns found:", list(df.columns))

# Ensure 'Stime' and 'Ltime' are present (or their UTC alternatives)
# If 'Stime_UTC' exists, we can use it or map it to 'Stime' if needed
display(df.head())
```

```python
# CELL 4: UNDERSTAND THE DATA
print("Data types:")
display(df.dtypes.to_frame("Data type"))
print("Missing values:")
display(df.isna().sum().sort_values(ascending=False).to_frame("Missing values"))
print("Label counts:")
display(df["Label"].value_counts(dropna=False).to_frame("Count"))
print("Attack category counts:")
display(df["attack_cat"].value_counts(dropna=False).to_frame("Count"))
```

```python
# CELL 5: BASIC CLEANING
clean_df = df.copy()

# Strip string columns
for col in clean_df.select_dtypes(include=["object"]).columns:
    clean_df[col] = clean_df[col].str.strip()

# Convert timestamp columns to numeric (handling Stime/Ltime or Stime_UTC/Ltime_UTC)
# If Stime_UTC and Ltime_UTC are datetime strings, convert them or ensure we use the numeric ones.
for col in ["Stime", "Ltime"]:
    if col in clean_df.columns:
        clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

# Label 0 means normal traffic; fill missing attack_cat only for these records.
if "Label" in clean_df.columns and "attack_cat" in clean_df.columns:
    normal_mask = clean_df["Label"].eq(0)
    clean_df.loc[normal_mask & clean_df["attack_cat"].isna(), "attack_cat"] = "Normal"

print("Missing values after handling normal labels:")
display(clean_df.isna().sum().sort_values(ascending=False).to_frame("Missing values"))
```

```python
# CELL 6: CHECK TIMESTAMPS AND DUPLICATES
for col in ["Stime", "Ltime", "Stime_UTC", "Ltime_UTC"]:
    if col in clean_df.columns:
        print(f"Rows with missing {col}:", clean_df[col].isna().sum())

print("Exact duplicate rows:", clean_df.duplicated().sum())

# Display timestamp range
active_time_col = "Stime_UTC" if "Stime_UTC" in clean_df.columns else "Stime"
if active_time_col in clean_df.columns:
    print(f"{active_time_col} range:", clean_df[active_time_col].min(), "to", clean_df[active_time_col].max())
```

```python
# CELL 7: SORT CHRONOLOGICALLY
# We will use Stime_UTC if available, or Stime as fallback
sort_cols = []
if "Stime_UTC" in clean_df.columns:
    sort_cols = ["Stime_UTC", "Ltime_UTC"]
else:
    sort_cols = ["Stime", "Ltime"]

chronological_all = clean_df.sort_values(
    by=sort_cols,
    ascending=[True, True],
    kind="mergesort",
    na_position="last"
).reset_index(drop=True)

valid_time = chronological_all[sort_cols[0]].dropna()
print("Chronological order verified:", valid_time.is_monotonic_increasing)

show_cols = [c for c in [sort_cols[0], sort_cols[1], "proto", "attack_cat", "Label"] if c in clean_df.columns]
display(chronological_all[show_cols].head(10))
```

```python
# CELL 8: CREATE DEDUPLICATED VERSION
chronological_deduplicated = chronological_all.drop_duplicates().reset_index(drop=True)
print("All-records version:", len(chronological_all))
print("Deduplicated version:", len(chronological_deduplicated))
print("Exact duplicates removed:", len(chronological_all) - len(chronological_deduplicated))
```

```python
# CELL 9: SAVE BOTH DATASETS
all_file = "UNSW_NB15_DriftWise_chronological_all_records.csv.gz"
dedup_file = "UNSW_NB15_DriftWise_clean_chronological.csv.gz"
chronological_all.to_csv(all_file, index=False, compression="gzip")
chronological_deduplicated.to_csv(dedup_file, index=False, compression="gzip")
print("Saved:", all_file, "and", dedup_file)
```

## Data checks: active window and normal-only tail
All attacks occur before a certain row. This cell cleans the ports, splits the file at the last attack, and shows how the attack rate and attack types change over time.

```python
import pandas as pd, numpy as np

path = "/content/drive/MyDrive/DriftWise/data/processed/UNSW_NB15_DriftWise_clean_chronological.csv.gz"
df = pd.read_csv(path, low_memory=False)

# Clean the two port columns: bad values (like 0x000b or -) become missing
for c in ["sport", "dsport"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
print("Missing ports after cleaning:", int(df["sport"].isna().sum()), "in sport,",
      int(df["dsport"].isna().sum()), "in dsport")

# Split at the last attack: everything before is the 'active window', the rest is normal-only
last_attack = int(np.where(df["Label"].values == 1)[0].max())
active = df.iloc[: last_attack + 1].copy()
tail = df.iloc[last_attack + 1 :].copy()
print("\nActive window:", len(active), "rows,", int(active["Label"].sum()), "attacks,",
      round(100 * active["Label"].mean(), 2), "% attack rate")
print("Normal-only tail:", len(tail), "rows,", int(tail["Label"].sum()), "attacks")

# Cut the active window into 10 equal chronological slices and describe each
active["slice"] = pd.qcut(np.arange(len(active)), 10, labels=False)
table = active.groupby("slice").agg(rows=("Label", "size"), attacks=("Label", "sum"))
table["attack_rate_%"] = (100 * table["attacks"] / table["rows"]).round(2)
top = active[active["Label"] == 1].groupby("slice")["attack_cat"].agg(
    lambda s: ", ".join(f"{k} {v}" for k, v in s.value_counts().head(3).items()))
table["top attack types"] = top
pd.set_option("display.max_colwidth", None, "display.width", 200)
display(table)
```
